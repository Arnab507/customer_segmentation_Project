```python
import pandas as pd
import numpy as np
```

```python
# Load dataset
df=pd.read_csv('online_retail.csv')
```

```python
# View dataset
df.head()
```

```python
df.tail()
```

```python
df.describe()
```

```python
#Check For null value
df.isnull().sum()
```

```python
# Remove rows with missing CustomerID
df = df.dropna(subset=['CustomerID'])
```

```python
df.isnull().sum()
```

```python
df.columns
```

```python
#Rename column
df=df.rename(columns={'index':'Index'})
```

```python
df.columns
```

```python
# Create TotalPrice column
df['TotalPrice'] = df['Quantity'] * df['UnitPrice']
```

```python
df.head()
```

```python
#Statistical Calculations
avg_sales = np.mean(df['TotalPrice'])

max_sales = np.max(df['TotalPrice'])

min_sales = np.min(df['TotalPrice'])


print("Average Sale: " ,avg_sales)
print("Maximum Sale:" ,max_sales)
print("Minimum Sale:" ,min_sales)
```

```python
# Convert InvoiceDate into datetime
df['InvoiceDate'] = pd.to_datetime(df['InvoiceDate'])

# Extract date features
df['Year'] = df['InvoiceDate'].dt.year
df['Month'] = df['InvoiceDate'].dt.month
df['Day'] = df['InvoiceDate'].dt.day
df['Hour'] = df['InvoiceDate'].dt.hour
```
